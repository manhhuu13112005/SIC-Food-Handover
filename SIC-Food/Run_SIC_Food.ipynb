# Khởi động SIC-Food
Chọn kernel sic-food. Chạy lần lượt; dừng server trước khi khởi động lại. Dấu hiệu bên ngoài không chứng nhận an toàn thực phẩm.

```python
import sys
from pathlib import Path
PROJECT = Path(r"C:\Users\ASUS\SIC-Food")
print("Python kernel:", sys.executable)
print("Project:", PROJECT)
assert (PROJECT / "main.py").is_file(), "Chưa thấy main.py trong PROJECT"
```

```python
import os
import subprocess
if "process" in globals() and process.poll() is None:
    raise RuntimeError("Server đang chạy; dùng cell dừng trước khi chạy lại")
env = os.environ.copy()
env["SIC_MODEL_PATH"] = str(PROJECT / "best.pt")
with (PROJECT / "streamlit.log").open("a", encoding="utf-8") as log:
    process = subprocess.Popen(
        [sys.executable, "-m", "streamlit", "run", str(PROJECT / "main.py"),
         "--server.address", "127.0.0.1", "--server.port", "8501"],
        cwd=str(PROJECT), env=env, stdout=log, stderr=subprocess.STDOUT,
        creationflags=subprocess.CREATE_NO_WINDOW)
print("PID:", process.pid, "— mở http://localhost:8501")
```

```python
# Kiểm tra server; None nghĩa là tiến trình vẫn đang chạy.
print("Exit code:", process.poll())
print((PROJECT / "streamlit.log").read_text(encoding="utf-8", errors="replace")[-4000:])
```

```python
# Chỉ chạy cell này khi muốn dừng ứng dụng.
if "process" in globals() and process.poll() is None:
    process.terminate()
    process.wait(timeout=15)
    print("Đã dừng Streamlit")
```
